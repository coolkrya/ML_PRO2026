```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('data/train.csv')
cols1 = data.columns
data = data.rename({'1stFlrSF': 'FstFlrSF', 
              '2ndFlrSF': 'SndFlrSF', 
              '3SsnPorch': 'ThirdSsnPorch'}, axis=1)
```

```python
data.columns
```

```python
categeorical_columns = [column for column in data.columns if data[column].dtype == 'str']
quantitive_columns = [column for column in data.columns if data[column].dtype != 'str' and column != "SalePrice"]
print('categorical_columns =', categeorical_columns)
print('quantitive_columns =', quantitive_columns)
```

```python
missing_values = data.isna().sum().sort_values(ascending=False)
missing_values_columns = missing_values[missing_values > 0]
missing_values_columns
```

```python
for i in categeorical_columns:
    if i not in missing_values_columns:
        print(i)
```

# Preprocessing test

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import TargetEncoder
```

```python
pd.DataFrame(TargetEncoder(target_type='continuous').fit_transform(data[categeorical_columns], data.iloc[::, -1]), columns=categeorical_columns)
```

```python
pd.DataFrame(StandardScaler().fit_transform(SimpleImputer(strategy='median').fit_transform(data[quantitive_columns])), columns=quantitive_columns)
```

```python
categorical_columns = [
    "MSZoning",
    "Street",
    "Alley",
    "LotShape",
    "LandContour",
    "Utilities",
    "LotConfig",
    "LandSlope",
    "Neighborhood",
    "Condition1",
    "Condition2",
    "BldgType",
    "HouseStyle",
    "RoofStyle",
    "RoofMatl",
    "Exterior1st",
    "Exterior2nd",
    "MasVnrType",
    "ExterQual",
    "ExterCond",
    "Foundation",
    "BsmtQual",
    "BsmtCond",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2",
    "Heating",
    "HeatingQC",
    "CentralAir",
    "Electrical",
    "KitchenQual",
    "Functional",
    "FireplaceQu",
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",
    "PavedDrive",
    "PoolQC",
    "Fence",
    "MiscFeature",
    "SaleType",
    "SaleCondition",
]
quantitive_columns = [
    "Id",
    "MSSubClass",
    "LotFrontage",
    "LotArea",
    "OverallQual",
    "OverallCond",
    "YearBuilt",
    "YearRemodAdd",
    "MasVnrArea",
    "BsmtFinSF1",
    "BsmtFinSF2",
    "BsmtUnfSF",
    "TotalBsmtSF",
    "FstFlrSF",
    "SndFlrSF",
    "LowQualFinSF",
    "GrLivArea",
    "BsmtFullBath",
    "BsmtHalfBath",
    "FullBath",
    "HalfBath",
    "BedroomAbvGr",
    "KitchenAbvGr",
    "TotRmsAbvGrd",
    "Fireplaces",
    "GarageYrBlt",
    "GarageCars",
    "GarageArea",
    "WoodDeckSF",
    "OpenPorchSF",
    "EnclosedPorch",
    "ThirdSsnPorch",
    "ScreenPorch",
    "PoolArea",
    "MiscVal",
    "MoSold",
    "YrSold",
]

numeric_transformer = Pipeline(steps=[("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())])


categorical_transformer = Pipeline(steps=[("target_enc", TargetEncoder(target_type="continuous", smooth="auto"))])


preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, quantitive_columns),
        ("cat", categorical_transformer, categeorical_columns),
    ],
    remainder="drop",
)
```

```python
preprocessor.fit_transform(data.iloc[::, :-1:], data.iloc[::, -1])
```

# Import preprocessing pipeline

```python
from preprocessing import preprocessor
```

# data split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    data.iloc[::, :-1:], data.iloc[::, -1], test_size=0.33, random_state=42)
```

```python
X_train.columns
```

# Model

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
model_pipeline = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('model', GradientBoostingRegressor())
])
```

```python
model_pipeline.fit(X_train, y_train)
```

```python
preds = model_pipeline.predict(X_test)
preds
```

```python
abs(np.log(y_test) - np.log(preds)).mean()
```

# Saving model

```python
import joblib

metadata = {
    'version': '1.0.0',
    'features': X_train.columns
}

bundle = {
    "pipeline": model_pipeline,
    "metadata": metadata
}

joblib.dump(bundle, "../artifact/model_1.0.joblib")
```
